Topik13 - List Comprehension
----------------------------

-> List Comprehension adalah cara singkat untuk membuat list baru berdasarkan list yang sudah ada

sintaks sederhana:
``` python
[ekspresi for item in iterable]
```

case:

Semisal saya punya list "angka" 1 sampai 5, dan saya ingin membuat list baru ("squares") yang berisi hasil pangkat dua dari list angka

``` python
# Menggunakan loop biasa
angka = [1, 2, 3, 4, 5]
squares = []
for num in angka:
    squares.append(num ** 2)

# Menggunakan List Comprehension
squares = [num ** 2 for num in angka]
```
Penggunaan list comprehension lebih ringkas dan cepat dari pada menggunakan for + append()

Cara lain menggunakan list comprehension:
``` python
[eksepresi for item in iterable if kondisi] # membuat list baru dengan teknik filter (if)
[ekspresi_A if kondisi else ekspresi_B for item in iterable] # membuat list baru berdasarkan kondisi tertentu 
```

Meski list comprehension ini terlihat ringkas namun tetap memiliki kekurangan dikondisi tertentu:
- saat logika terlalu kompleks sehingga kode akan sulit dibaca
- saat memproses data raksasa

````python
squares = [x**2 for x in range(10)]
print(squares)
````

````python
numbers = [1,2,3,4,5]
doubled = [x*2 for x in numbers]
print(doubled)
````

````python
numbers = [1,2,3,4,5,6,7]
evens = [x for x in numbers if x % 2 == 0]
print(evens)
````

````python
names = ["Alice", "Bob", "Charlie", "Dave"]
short_names = [name for name in names if len(name) < 5]
print(short_names)
````

````python
numbers = [1,2,3,4,5,6]
labels = ["Even" if x % 2 == 0 else "Odd" for x in numbers]
print(labels)
````

````python
student_scores = input("Enter student scores separated by commas: ")
scores = [int(score) for score in student_scores.split(",")]

grades = [
    "A" if score >= 90 else
    "B" if score >= 80 else
    "C" if score >= 70 else
    "D" if score >= 60 else
    "F"
    for score in scores
]

passing_student = [score for score in scores if score >= 60]
failing_student = [score for score in scores if score < 60]

print("\n--- Student Grades")
for i, (score, grade) in enumerate(zip(scores, grades), start=1):
    print(f"Student {i}: Score = {score}, Grade = {grade}")

print("\n--- Passing and Failing Students ---")
print("Passing students:", passing_student)
print("Failing students:", failing_student)
````

## Tugas:
Buatlah satu baris List Comprehension untuk menghasilkan list email baru dengan ketentuan:
1. Email harus dibersihkan dari spasi liar di awal/akhir (strip).
2. Semua huruf diubah menjadi huruf kecil (lower()).
3. Hanya ambil email yang valid, yaitu yang mengandung domain @gmail.com atau @yahoo.com.

````python
raw_emails = ["  BUDI@gmail.com ", "siti@yahoo.com", "INVALID_EMAIL", "   andi@outlook.com  ", "admin@company.id"]

new_email = [email.strip().lower() for email in raw_emails if "@gmail.com" in email or "@yahoo.com" in email ]
print(new_email)
````

## Tugas:
Buatlah satu baris List Comprehension untuk memberikan label diskon berdasarkan aturan berikut:

1. Jika harga di atas atau sama dengan Rp 200.000, berikan diskon 20% (dikalikan 0.8).
2. Jika harga di atas atau sama dengan Rp 100.000 (dan di bawah 200.000), berikan diskon 10% (dikalikan 0.9).
3. Jika harga di bawah Rp 100.000, harga tetap/tidak ada diskon (dikalikan 1.0).

````python
prices = [45000, 120000, 250000, 80000, 300000]

after_discount = [price*0.8 if price >= 200000 else price*0.9 if price >= 100000 else price for price in prices]
print(after_discount)
````

## Tugas:
Buatlah satu baris List Comprehension untuk mengekstrak string ringkasan dengan format "NAMA - Rp TOTAL" hanya untuk transaksi yang statusnya "completed" dan total belanjanya di atas Rp 100.000.

````python
transactions = [
    {"user": "Budi", "items": ["Buku", "Pensil"], "total": 45000, "status": "completed"},
    {"user": "Siti", "items": ["Laptop"], "total": 8500000, "status": "pending"},
    {"user": "Andi", "items": ["Mouse", "Keyboard"], "total": 350000, "status": "completed"},
    {"user": "Eko", "items": [], "total": 0, "status": "failed"},
    {"user": "Dewi", "items": ["Monitor"], "total": 1500000, "status": "completed"}
]

ringkasan = [f"{trx['user']} - {trx['total']}" for trx in transactions if trx["status"] == "completed" and trx["total"] > 100000]
print(ringkasan)
````
