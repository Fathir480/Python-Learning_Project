```python
import csv

with open('laporan_tugas.csv', 'w', newline='', encoding='utf-8') as file_w, \
open('catatan_harian.txt', 'r', encoding='utf-8') as file_r:
    read = file_r.readlines()
    reading = [line.strip().split(', ') for line in read]
    header = ['Nama_Tugas', 'Status', 'Prioritas']
    dict_data = []
    for line in reading:
        baris_dict = {
            header[0]: line[0],
            header[1]: line[1],
            header[2]: line[2]
        }
        dict_data.append(baris_dict)

    writer = csv.DictWriter(file_w, fieldnames=header)
    writer.writeheader()
    writer.writerows(dict_data)
```
