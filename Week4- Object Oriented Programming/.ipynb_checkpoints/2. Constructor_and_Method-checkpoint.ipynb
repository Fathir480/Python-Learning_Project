# Constructor & Method

```
Construktor adalah sebuah special method yang ada di dalam kelas, yang otomatis dipanggil pertama kali saat kamu membuat sebuah objek (instansiasi).
- Di Python, constructor utama bernama __init__.
- Fungsi Utama: Untuk menyiapkan kondisi awal (initial state) sebuah objek, seperti memberikan nilai atribut pertama kali saat objek lahir di memori
```

````python
# __Contoh Constructor Sederhana:__
class Mahasiswa:
    def __init__(self, nama, nim, ipk):
        self.nama = nama
        self.nim = nim
        self.ipk = ipk

# instance mhs1 memiliki nilai atribut awal:
mhs1 = Mahasiswa("Ahmad Fathir", "13020240046", 3.98)

print(f"Nama: {mhs1.nama}, NIM: {mhs1.nim}, IPK: {mhs1.ipk}")
````

# Jenis-jenis Method dalam OOP Python
method dibagi menjadi 3 jenis utama:

### A. Instance Method (Method Objek)
- Ciri khas: Menggunakan parameter pertama __self__
- Fungsi: Untuk mengakses atau memodifikasi atribut yang dimiliki oleh objek / instance tertentu

### B. Class Method (Method Kelas)
- Ciri khas: Menggunakan dekorator __@classmethod__ dan parameter __cls__
- Fungsi: Berinteraksi langsung dengan kelas itu sendiri, bukan objek individual. Biasanya dipakai untuk Alternative Construktor (membuat objek dari format data lain seperti JSON atau teks string)

### C. Static Method (Method Statis)
- Ciri khas: Menggunakan dekorator __@staticmethod__ tanpa parameter self maupun cls
- Fungsi: Seperti fungsi biasanya yang ditulis dalam class karena masih berkaitan secara logika. Tidak mengenal objek maupun kelas, dan method ini dapat digunakan tanpa melakukan instansiasi terlebih dahulu

````python
# Contoh Study Kasus: Sistem Akademik Sederhana
class Mahasiswa:
    # construktor utama
    def __init__(self, nama, ipk):
        self.nama = nama
        self.ipk = ipk
    
    # instance method
    def evaluasi_predikat(self):
        if self.ipk >= 3.50:
            return f"{self.nama} lulus dengan Cum Laude!"
        return f"{self.nama} lulus dengan baik."

    # class method
    @classmethod 
    def from_string(cls, teks_data):
        nama_mhs, ipk_mhs = teks_data.split("-")
        return cls(nama_mhs, ipk_mhs)

    # static method
    @staticmethod
    def validasi_ipk(ipk):
        if ipk < 0.0 or ipk > 4.0:
            return f"{ipk} tidak valid!"
        else:
            return f"{ipk} valid!"

mhs1 = Mahasiswa("Fathir", 3.75)
print(mhs1.evaluasi_predikat())

data_mentah = "Liona-3.20"
mhs2 = Mahasiswa.from_string(data_mentah)
print(mhs2.evaluasi_predikat)

cek1 = Mahasiswa.validasi_ipk(3.80)
cek2 = Mahasiswa.validasi_ipk(4.10)
print(cek1)
print(cek2)
````

### Studi Kasus 1: Sistem E-Commerce dengan @classmethod (Alternative Constructor)
Bayangkan kamu sedang membangun sistem untuk mengelola data produk.

- Buat sebuah kelas bernama Produk yang memiliki constructor utama (__init__) untuk menerima parameter nama, harga (integer), dan stok (integer).
- Buatlah sebuah @classmethod bernama dari_csv_string(cls, teks_csv) yang berfungsi sebagai Alternative Constructor. Method ini bertugas menerima string berformat "NamaProduk,Harga,Stok" (contoh: "Mechanical Keyboard,850000,15"), memecahnya, mengonversi tipe datanya, lalu mengembalikan objek Produk tersebut.
- Buat juga sebuah Instance Method bernama cek_ketersediaan(cls) atau tampilkan_detail(self) untuk mencetak informasi produk.
- Uji kodenya dengan membuat satu objek via constructor biasa, dan satu objek lagi via Alternative Constructor dari_csv_string.

````python
class Produk:
    def __init__(self, nama, harga, stok):
        self.nama = nama
        self.harga = int(harga)
        self.stok = int(stok)

    @classmethod
    def dari_csv_string(cls, teks_csv):
        nama, harga, stok = teks_csv.split(",")
        return cls(nama, harga, stok)

    def cek_ketersediaan(self):
        print(f"Produk {self.nama} tersedia {self.stok} pcs")

    def tampilkan_detail(self):
        print(f"Nama Produk : {self.nama}")
        print(f"Harga       : {self.harga}")
        print(f"Stok        : {self.stok}")

    @staticmethod
    def hitung_harga_setelah_diskon(harga, persentase):
        total = harga * (1 - persentase/100)
        return total

produk1 = Produk("Mouse", 300000, 30)

teks = "Keyboard,500000,15"
produk2 = Produk.dari_csv_string(teks)

produk1.tampilkan_detail()
produk2.cek_ketersediaan()

harga_total = Produk.hitung_harga_setelah_diskon(500000, 20)
print(harga_total)
````
