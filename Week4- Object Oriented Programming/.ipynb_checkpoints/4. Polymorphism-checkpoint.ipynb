# Polymorphism

Polymorphism adalah konsep OOP di mana sebuah nama method atau fungsi yang sama dapat digunakan untuk melakukan tugas yang berbeda-beda, tergantung pada objek apa yang memanggilnya atau meresponnya

```python
# Contoh Kasus: Sistem Pembayaran
class MetodePembayaran:
    def proses_bayar(self, jumlah):
        raise NotImplementedError("Subclass harus mengimplementasikan method ini!")

class KartuKredit(MetodePembayaran):
    def proses_bayar(self, jumlah):
        print(f"Memproses pembayaran Rp {jumlah:,} menggunakan Kartu Kredit (Potong Limit Bank).")

class EWallet(MetodePembayaran):
    def proses_bayar(self, jumlah):
        print(f"Memproses pembayaran Rp {jumlah:,} menggunakan E-Wallet (Scan QR Code).")

class TransferBank(MetodePembayaran):
    def proses_bayar(self, jumlah):
        print(f"Memproses pembayaran Rp {jumlah:,} via Virtual Account Bank.")

# --- FUNGSI POLIMORFIK (Menerima objek apa saja, punya method proses_bayar) ---
def bayar_pesanan(alat_pembayaran, total_harga):
    alat_pembayaran.proses_bayar(total_harga)

p1 = KartuKredit()
p2 = EWallet()
p3 = TransferBank()

# Memanggil fungsi yang sama dengan objek yang berbeda-beda
bayar_pesanan(p1, 150000)
bayar_pesanan(p2, 45000)
bayar_pesanan(p3, 500000)
```

Dalam Python, jika kita ingin menerapkan polymorphism kita tidak wajib melakukan inheritance. Polymorphism dapat digunakan selama kelas memiliki nama method yang sama

```python
class Kucing:
    def bersuara(self):
        return "Meow! 🐱"

class Anjing:
    def bersuara(self):
        return "Guk Guk! 🐶"

class Robot:
    def bersuara(self):
        return "BEEP BOOP! 🤖"


def cetak_suara_hewan(objek_makhluk):
    print(objek_makhluk.bersuara())

hewan1 = Kucing()
hewan2 = Anjing()
mesin = Robot()

cetak_suara_hewan(hewan1)
cetak_suara_hewan(hewan2)
cetak_suara_hewan(mesin)
```
