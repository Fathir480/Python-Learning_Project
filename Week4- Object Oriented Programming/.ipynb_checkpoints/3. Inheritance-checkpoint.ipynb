# Inheritance (Pewarisan)

Inheritance adalah konsep di mana sebuah kelas baru (childclass) dapat mewarisi atribut dan method dari kelas yang sudah ada (superclass)

### Overriding method

overriding merupakan sebuah konsep dimana jika membuat method di childclass dengan nama method yang sama dengan di superclassnya maka method di childclass akan menimpa method di superclassnya

### Fungsi super()
super() bertugas memanggil konstruktor atau method milik superclassnya (kelas induknya)

```python
class Kendaraan:
    def __init__(self, merk, kecepatan):
        self.merk = merk
        self.kecepatan = kecepatan

    def bergerak(self):
        return f"Kendaraan {self.merk} sedang bergerak di jalan"

# membuat childclass
class Mobil(Kendaraan):
    def __init__(self, merk, kecepatan, tahun_produksi):
        super().__init__(merk, kecepatan)
        self.tahun_produksi = tahun_produksi
    
    # contoh overriding
    def bergerak(self):
        return f"Mobil {self.merk} melaju kencang dengan kecepatan {self.kecepatan} km/jam!"

kendaraan_umum = Kendaraan("Generic", 60)
mobil_sport = Mobil("Ferrari", 250, 1970)

print(kendaraan_umum.bergerak())
print(mobil_sport.bergerak())

print(mobil_sport.merk) # atribut yang di warisi
print(mobil_sport.tahun_produksi) # atribut khusus
```

### MRO (Method Resolution Order)

ketika kita ingin membuat class yang mewarisi beberapa kelas sekaligus

```python
class Pengguna:
    def akses_sistem(self):
        print("-> [1] Pengguna: Memverifikasi identitas dasar...")


class Karyawan(Pengguna):
    def akses_sistem(self):
        print("-> [2] Karyawan: Memberikan akses ke modul absensi & gaji.")
        super().akses_sistem()


class AdminIT(Pengguna):
    def akses_sistem(self):
        print("-> [3] AdminIT: Memberikan akses ke server & database.")
        super().akses_sistem()


class SuperAdmin(Karyawan, AdminIT):
    def akses_sistem(self):
        print("-> [4] SuperAdmin: Membuka panel kontrol utama tingkat tertinggi!")
        super().akses_sistem()


s_admin = SuperAdmin()
s_admin.akses_sistem()

# --- MENGECEK URUTAN MRO ---
print("\n[INFO] Jalur MRO untuk kelas SuperAdmin:")
for i, cls in enumerate(SuperAdmin.__mro__):
    print(f"{i+1}. {cls.__name__}")
```
