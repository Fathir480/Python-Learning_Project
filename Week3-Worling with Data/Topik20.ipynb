# Dates & Time in Python

di materi ini akan membahas terkait pengolahan data tanggal dan waktu di Python.

### datetime module
__fungsi yang sering digunaakan:__
1. date() : mengolah tahun, bulan, dan hari
2. time() : mengolah jam, menit, detik, dan mikrosekon
3. datetime() : mengolah kombinasi lengkap antara tanggal dan waktu
4. timedelta() : mengolah selisih atau durasi waktu

```python
from datetime import date, time
tgl_ulang_tahun = date(2006, 5, 3) # contoh date()
print(tgl_ulang_tahun)

waktu_tidur = time(9, 30, 0) # contoh time()
print(waktu_tidur)
```

### Cara mengambil waktu / tanggal saat ini

```python
from datetime import datetime, date
# Mengambil tanggal dan waktu saat ini
sekarang = datetime.now()
print("Waktu sekarang:", sekarang)

# Mengambil komponen spesifiknya saja
print("Tahun :", sekarang.year)
print("Bulan :", sekarang.month)
print("Hari  :", sekarang.day)
print("Jam   :", sekarang.hour)
print("Menit :", sekarang.minute)

# Mengambil tanggal hari ini
hari_ini = date.today()
print("Tanggal Hari ini:", hari_ini)
```

### Cara konversi string ke tanggal dan tanggal ke string

Tabel Kode Format Penting (strftime / strptime):
- %d: Hari (01-31)
- %m: Bulan angka (01-12)
- %B: Nama bulan lengkap (August, December)
- %b: Nama bulan singkat (Aug, Dec)
- %Y: Tahun 4 digit (2026)
- %H: Jam format 24 jam (00-23)
- %M: Menit (00-59)
- %S: Detik (00-59)

```python
# String ke datetime
from datetime import datetime

string_tanggal = "05-08-2026 14:30:00"

tanggal = datetime.strptime(string_tanggal, "%d-%m-%Y %H:%M:%S") # strptime - Parse Time

print("Tipe data:", type(tanggal))
print("Tahun hasil parse:", tanggal.year)
```

```python
# Datetime ke string
from datetime import datetime

sekarang = datetime.now()

format_rapi = sekarang.strftime("%d %B %Y, Jam %H:%M WIB") # strftime - Format Time
print("Hasil Format:", format_rapi)
```

### Operasi Waktu dan Selisih Hari

- datetime merupakan typedata yang memiliki format tanggal
- timedelta merupakan typedata yang bermakna berapa lama atau durasi

operasi waktu:
- datetime $-$ datetime = timedelta
- datetime $\pm$ timedelta = datetime

bentuk timedelta = hari, waktu [148 days, 20:20:15] artinya, 148 hari 20 jam 20 mnt dan 15 detik
- .days untuk mengambil harinya saja [148]
- .seconds untuk mengambil waktunya saja dalam satuan detik [73092]
- .total_seconds() untuk mengambil hari + waktu dalam satuan detik []

```python
from datetime import datetime, timedelta

sekarang = datetime.now()

tujuh_hari_lagi = sekarang + timedelta(days=7) # contoh timedelta
dua_jam_lalu = sekarang - timedelta(hours=2)

print("7 Hari Lagi:", tujuh_hari_lagi.strftime("%d-%m-%Y"))
print("2 Jam Lalu :", dua_jam_lalu.strftime("%H:%M"))

tahun_baru = datetime(2027, 1, 1)
selisih = tahun_baru - sekarang
print(f"\nBentuk timedelta: {selisih}")
print(f"Mengambil hari saja: {selisih.days}")
print(f"Mengambil waktu saja: {selisih.seconds}")
print(f"Mengambil total hari dan waktu: {selisih.total_seconds()}")

print(f"\nSisa waktu menuju tahun baru: {selisih.days} hari, {selisih.seconds // 3600} jam lagi!")
```

### Studi Kasus 1: Hitung Usia & Ulang Tahun Berikutnya (Basic Logic & Date)
__Skenario:__

Anda diminta membuat program yang menghitung total usia seseorang dalam hari dan menghitung sisa hari menuju ulang tahun berikutnya.

__Tugas Anda:__

1. Minta input tanggal lahir pengguna dalam format YYYY-MM-DD (misal: "2006-03-05").
2. Ubah string tersebut menjadi objek date menggunakan datetime.strptime().date().
3. Hitung berapa total hari yang sudah dilewati pengguna dari lahir sampai date.today().
4. Hitung berapa sisa hari menuju ulang tahun berikutnya di tahun ini (jika sudah lewat di tahun ini, hitung untuk tahun depan).

```python
from datetime import datetime, date

tanggal_lahir = input("Masukkan tanggal lahir Anda ['YYYY-MM-DD']: ")
format_tgl_lahir = datetime.strptime(tanggal_lahir, '%Y-%m-%d').date()

hari_ini = date.today()
total_hari =  (hari_ini - format_tgl_lahir).days

ulang_tahun = date(hari_ini.year+1, format_tgl_lahir.month, format_tgl_lahir.day)
sisa_hari = (ulang_tahun - hari_ini).days

print("\n---------------")
print(f"Tanggal lahir: {format_tgl_lahir}")
print(f"Total hari: {total_hari}")
print(f"Sisa hari: {sisa_hari}")
```

### Studi Kasus 2: Sistem Kasir & Masa Kedaluwarsa Diskon (Timedelta & Conditional)
__Skenario:__

Sebuah toko memberikan kupon diskon yang hanya berlaku selama 3 hari 12 jam dari waktu transaksi.

__Tugas Anda:__

1. Ambil waktu transaksi saat ini menggunakan datetime.now().
2. Hitung waktu batas kadaluwarsa diskon (waktu_transaksi + timedelta(...)).
3. Buat simulasi batas waktu klaim kupon dari pengguna (misal buat variabel waktu_klaim sembarang).
4. Buat logika if-else:
   - Jika waktu_klaim <= batas_kadaluwarsa, cetak: Diskon berhasil digunakan!.
   - Jika tidak, cetak: Diskon sudah kadaluwarsa!.
5. Cetak pula tanggal batas kadaluwarsa dalam format rapi: DD Month YYYY, Jam HH:MM WIB (contoh: 08 August 2026, Jam 16:30 WIB).

```python
from datetime import datetime, time, date

waktu_trx = datetime.now()
batas_waktu_kupon = waktu_trx + timedelta(days=3,hours=12)
waktu_claim = datetime(2026, 9, 6, 0, 0, 0)

if waktu_claim <= batas_waktu_kupon:
    print("Diskon berhasil digunakan")
else:
    print("Diskon sudah kadaluarsa!")
    print(batas_waktu_kupon.strftime("%d %B %Y, Jam %H:%M WIB"))
```

5. Tampilkan format waktu lokal kota tersebut dalam bahasa yang mudah dibaca.

```python
import requests
import json
from datetime import datetime

url = 'http://api.weatherapi.com/v1/current.json'
api_key = '86412f801a0c4053893194448260208'
kota = "London"
params = {
    'key': api_key,
    'q': kota
}

response = requests.get(url, params=params)
if response.status_code == 200:
    data = response.json()
    localtime = data['location']['localtime']
else:
    print("Data tidak ditemukan")

str_localtime = datetime.strptime(localtime, "%Y-%m-%d %H:%M")
sekarang = datetime.now()

selisih = sekarang - str_localtime
total_detik = selisih.total_seconds()

print(f"\n--- Perbandingan Waktu Lokal vs {kota} ---")
print(f"Waktu Lokal     : {sekarang.strftime('%Y-%m-%d %H:%M')}")
print(f"Waktu di {kota} : {localtime}")
print(f"Beda Waktu      : {total_detik // 3600:.0f} Jam")
```
