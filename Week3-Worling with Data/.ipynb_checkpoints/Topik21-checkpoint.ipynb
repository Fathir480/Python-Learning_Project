# WEB Scraping

Web scraping adalah teknik untuk mengambil data secara otomatis dari halaman web (HTML) menggunakan kode python.
Web scraping digunakan ketika situs web tidak menyediakan api resmi, sehingga kita harus membaca kode HTML web tersebut lalu menyaring informasi yang di butuhkan

__Tools Utama yang digunakan:__
1. requests
2. Beautifulsoup

__Fungsi yang sering di gunakan__
- BeautifulSoup(html_text, 'html.parser')  |  Mengubah string HTML mentah menjadi objek BeautifulSoup yang bisa dicari
- soup.find('tag', class_='...')  |  mencari 1 elemen pertama yang cocok dengan kriteria
- soup.find_all('tag', class_='...')  |  mencari semua elemen yang cocok (mengembalikan list)


```python
import requests
from bs4 import BeautifulSoup

url = "http://quotes.toscrape.com/"
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36'
}
response = requests.get(url, headers=headers)

if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')
    daftar_kutipan = soup.find_all('div', class_='quote')

    for q in daftar_kutipan:
        teks = q.find('span', class_='text').text
        penulis = q.find('small', class_='author').text
        
        print(f"Quote : {teks}")
        print(f"Author: {penulis}")
        print("-" * 40)
else:
    print("Gagal mengambil halaman web")
```

```python
# latihan scraping http://books.toscrape.com/
import requests
import csv
from bs4 import BeautifulSoup

url = 'http://books.toscrape.com/'
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/147.0.0.0 Safari/537.36'
}
harga_buku = []
response = requests.get(url, headers=headers)
if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')
    semua_buku = soup.find_all('article', class_='product_pod')

    for buku in semua_buku: 
        judul = buku.find('h3').find('a')['title']
        price = buku.find('p', class_='price_color').text

        dict_buku = {
            "Judul": judul,
            "Harga": price
        }
        harga_buku.append(dict_buku)
else:
    print("Data tidak ditemukan")

with open('daftar_harga_buku.csv', 'w', newline='', encoding='utf-8') as file:
    fieldnames = ["Judul","Harga"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(harga_buku)
```
