# JSON File in Python

JSON adalah format penyimpanan data berbasis teks yang strukturnya sangat mirip dengan Dictionary di Python (menggunakan pasangan key-value).

### Menulis File JSON

```python
import json

hasil_analisis = {
    "model": "RandomForest",
    "akurasi": 0.95,
    "status": "Ready to deploy" 
}

with open('hasil_model.json', 'w', encoding='utf-8') as file:
    json.dump(hasil_analisis, file, indent=4)
```

### Membaca File JSON

```python
import json

with open('hasil_model.json', 'r', encoding='utf-8') as file:
    data = json.load(file)
    print(data)
    print(data['akurasi'])
```

### Function umum yang di gunakan dari modul json

1. __json.loads()__: mengubah string berformat JSON menjadi dictionary
2. __json.dumps()__: mengubah dictionary menjadi string berformat JSON
3. __json.load()__ : membaca file .json dan mengubahnya menjadi dictionary
4. __json.dump()__ : menulis atau menyimpan dictionary ke file .json  

### json.loads() vs json.dumps()

File json itu ditulis dalam bentuk string of dict:
- contoh yang benar = '{"nama": "Ahmad"}'
- contoh yang salah = "{'nama': 'Ahmad'}"

```python
import json

data_string = '{"nama": "Ahmad", "usia": 20}' # Format asli JSON

data_python = json.loads(data_string)

print(type(data_python))
print(data_python['nama'])
```

```python
import json

data_python = {"nama": "Ahmad", "usia": 20}

string_json = json.dumps(data_python)

print(type(string_json))
print(string_json)
```
