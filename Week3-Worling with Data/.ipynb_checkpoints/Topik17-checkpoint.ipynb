# CSV Files

### Apa itu File CSV
__CSV__ adalah file teks biasa dimana setiap baris mewakili satu row (baris data), dan setiap kolomnya dipisahkan oleh karakter pemisah (delimiter), berupa koma (,) atau titik koma (;).

__Kenapa penting untuk Data/ML?__ Karena formatnya ringan, bisa dibuka di Excel maupun dibaca langsung oleh library Python seperti csv, pandas, hingga numpy tanpa butuh software berat.

### Menulis CSV menggunakan modul csv

````python
# Menulis menggunakan csv.writer
import csv

data1 = [
    ['Nama', 'Usia', 'Kota'],
    ['Ahmad', '20', 'Makassar'],
    ['Budi', '22', 'Jakarta']
]

with open('data.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerows(data1)
````

````python
# Menulis menggunakan csv.DictWriter (Berbasis Dictionary)
import csv

data2 = [
    {'Nama': 'Ahmad', 'Usia': 20},
    {'Nama': 'Budi', 'Usia': 22}
]

with open('data_dict.csv', 'w', newline='', encoding='utf-8') as file:
    fieldnames = ['Nama', 'Usia']
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    
    writer.writeheader() # Header 
    writer.writerows(data2) # Isi data
````

### Membaca CSV menggunakan modul csv

````python
# Membaca sebagai List (csv.reader)
import csv

with open('data.csv', 'r', encoding='utf-8') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
# Setiap row akan berupa list string
````

````python
# Membaca sebagai Dictionary (csv.DictReader)
import csv

with open('data_dict.csv', 'r', encoding='utf-8') as file:
    reader = csv.DictReader(file)
    for row in reader:
        print(row['Nama'], row['Usia']) # akses langsung dengan nama kolom sebagai key
````

# Studi Kasus: Analisis Penjualan E-Commerce Sederhana
Bayangkan Anda bekerja sebagai Data Analyst di sebuah toko online. Anda diberikan file data transaksi kotor dalam bentuk CSV dan diminta untuk membuat program Python yang otomatis merapikan, memproses, dan mencari wawasan (insights) dari data tersebut.

### 1. Buat File Dataset Dummy Terlebih Dahulu (transaksi.csv)
Pertama, buat file bernama transaksi.csv di folder proyek Anda menggunakan Python (atau tulis manual) dengan isi seperti ini:

```
id_transaksi,nama_produk,kategori,harga_satuan,jumlah_terjual
101,Laptop Asus,Elektronik,7500000,2
102,Kopi Arabika 250g,F&B,75000,10
103,Mouse Wireless,Elektronik,150000,5
104,Kopi Robusta 500g,F&B,90000,8
105,Mechanical Keyboard,Elektronik,500000,3
106,Tumbler Stainless,Lifestyle,120000,4
```

### 2. Tantangan Coding untuk Anda:
Tulislah sebuah program Python menggunakan modul csv (disarankan menggunakan csv.DictReader) yang melakukan tugas-tugas berikut:

- Membaca dan Menampilkan Data: Baca file transaksi.csv dan cetak nama produk beserta kategori dari setiap baris dengan format yang rapi (misal: - Laptop Asus (Kategori: Elektronik)).
- Kalkulasi Total Pendapatan per Baris: Tambahkan perhitungan di dalam loop untuk mencari Total Pendapatan dari setiap transaksi (Rumus: harga_satuan * jumlah_terjual), lalu cetak hasilnya.
- Pencarian Berdasarkan Kategori (Filter): Buat sebuah fungsi sederhana yang menerima argumen kategori_pilihan (misalnya "Elektronik"), lalu program hanya menampilkan produk-produk yang masuk ke dalam kategori tersebut beserta total pendapatan penjualannya.
- Menyimpan Laporan Baru (laporan_pendapatan.csv): Buat file CSV baru bernama laporan_pendapatan.csv menggunakan csv.DictWriter yang berisi data aslinya ditambah satu kolom baru yaitu total_pendapatan.

````python
import csv

data = [
    ['id_transaksi','nama_produk','kategori','harga_satuan','jumlah_terjual'],
    [101,'Laptop Asus','Elektronik',7500000,2],
    [102,'Kopi Arabika 250g','F&B',75000,10],
    [103,'Mouse Wireless','Elektronik',150000,5],
    [104,'Kopi Robusta 500g','F&B',90000,8],
    [105,'Mechanical Keyboard','Elektronik',500000,3],
    [106,'Tumbler Stainless','Lifestyle',120000,4]
]

with open('transaksi.csv', 'w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    writer.writerows(data)
````

````python
def baca_data(file_name):
    with open(file_name, 'r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        for row in reader:
            print(f"- {row['nama_produk']} ({row['kategori']})")

def hitung_total(file_name):
    with open(file_name, 'r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        for row in reader:
            total = int(row['harga_satuan']) * int(row['jumlah_terjual'])
            print(f"- {row['nama_produk']} : {total}")

def cari(file_name, kategori):
    with open(file_name, 'r', encoding='utf-8') as file:
        reader = csv.DictReader(file)
        for row in reader:
            if kategori == row['kategori']:
                total = int(row['harga_satuan']) * int(row['jumlah_terjual'])
                print(f"- {row['nama_produk']} : {total}")

def buat_laporan_pendapatan(file_asal, file_tujuan):
    with open(file_asal, 'r', encoding='utf-8') as f_baca, \
         open(file_tujuan, 'w', newline='', encoding='utf-8') as f_tulis:

        reader = csv.DictReader(f_baca)
        
        fieldnames_asli = reader.fieldnames
        fieldnames_baru = fieldnames_asli + ['total_pendapatan']
    
        writer = csv.DictWriter(f_tulis, fieldnames=fieldnames_baru)
        writer.writeheader()
        
        for row in reader:
            pendapatan = int(row['harga_satuan']) * int(row['jumlah_terjual'])
            row['total_pendapatan'] = pendapatan
            writer.writerow(row)

baca_data('transaksi.csv')

print('\nPendapatan tiap produk')
hitung_total('transaksi.csv')

print('\nProduk dengan katefori Elektronik')
cari('transaksi.csv', 'Elektronik')
buat_laporan_pendapatan('transaksi.csv', 'laporan_pendapatan.csv')
````
