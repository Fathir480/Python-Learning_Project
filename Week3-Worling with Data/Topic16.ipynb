# Writing Files in Python

sama seperti topic sebelumnya, kita masih akan menggunakan fungsi open() namun dengan mode yang berbeda

### mode 'w' vs. mode 'a'

__'w'__ atau Write:
- Jika file belum ada, Pyhton akan membuatkan file baru
- Jika file sudah ada, Python akan menghapus seluruh isi didalam file dan menggantinya dengan tulisan yang baru

__'a'__ atau Append:
- Jika file belum ada, Pyhton akan membuatkan file baru
- JIka file sudah ada, Python akan menambahkan tulisan baru di baris akhir

Contoh sintaks:

```python
# 'w'
with open('catatan.txt', 'w', encoding='utf-8') as file:
    file.write("Halo, ini adalah baris pertama.\n")
    file.write("Ini baris kedua.\n")
```

```python
# 'a'
with open('catatan.txt', 'a', encoding='utf-8') as file:
    file.write("Ini baris tambahan yang baru dimasukkan.\n")
```

### write() vs. writelines()

- fungsi __write()__ digunakan untuk menulis dalam satu baris
- fungsi __writelines()__ menerima input list yang dapat menjadikannya beberapa baris

```python
daftar_buah = ["Apel\n", "Jeruk\n", "Mangga\n"]

with open('buah.txt', 'w', encoding='utf-8') as file:
    file.writelines(daftar_buah)
```

## Studi Kasus 1: Program Buku Tamu Sederhana (Mode a)
Bayangkan Anda sedang membuat aplikasi kecil untuk mencatat nama tamu yang datang ke sebuah acara.

Tantangan:
- Buat sebuah fungsi bernama tambah_tamu(nama_file, nama_tamu).
- Fungsi ini harus menambahkan (append) nama tamu yang diinput ke dalam file teks (misalnya tamu.txt).
- Pastikan setiap nama tamu yang ditambahkan berada di baris baru (tidak menempel dengan nama sebelumnya).

Uji fungsi tersebut dengan memasukkan beberapa nama secara berurutan.

```python
def tambah_tamu(nama_file, nama_tamu):
    with open(nama_file, 'a', encoding='utf-8') as file:
        file.write(nama_tamu + '\n')

print("--- Sistem Buku Tamu Sederhana ---")
input_tamu = input("Masukkan nama tamu yang datang: ")
tambah_tamu('tamu.txt', input_tamu)
```

## Studi Kasus 2: Sistem Pencatatan Skor Game (Mode w vs a)
Bayangkan Anda membuat game sederhana dan ingin menyimpan High Score pemain.

Tantangan:
- Buat sebuah program yang meminta pengguna memasukkan nama pemain dan skor mereka.
- Simpan data tersebut ke dalam file highscore.txt.
- Syarat khusus: Pikirkan baik-baik, mode apa ('w' atau 'a') yang harus Anda gunakan jika file highscore.txt ini hanya boleh menyimpan satu rekor skor terbaru saja (selalu ditimpa setiap kali ada permainan baru)? Tuliskan kodenya.

```python
nama = input("Masukkan nama pemain: ")
score = input("Masukkan scorenya: ")

teks = f"{nama}: {score}"
with open("highscore.txt", "w", encoding="utf-8") as file:
    file.write(teks)
    
```
