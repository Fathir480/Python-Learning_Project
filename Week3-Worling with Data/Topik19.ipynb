# Apa itu API?

API adalah singkatan dari Application Programming Interface

Secara sederhana, API adalah "jembatan" atau "pelayan" yang menghubungkan aplikasi Anda dengan sistem atau server lain, sehingga aplikasi Anda bisa meminta data atau layanan tanpa perlu tahu bagaimana server tersebut dibangun di dalamnya.

```
[ Program Python ] ----( 1. Request / Minta Data )----> [ Server API ]
                   <---( 2. Response / Mengembalikan JSON )--
```


````python
# contoh mengambil data dengan api
import requests

url = "https://jsonplaceholder.typicode.com/users/1"
response = requests.get(url)
if response.status_code == 200:
    data = response.json()
    
    print("--- Data Berhasil Diambil ---")
    print(f"Nama     : {data['name']}")
    print(f"Email    : {data['email']}")
    print(f"Kota     : {data['address']['city']}")
else:
    print(f"Gagal mengambil data. Status code: {response.status_code}")
````

````python
import requests

api_key = '86412f801a0c4053893194448260208'
url = "http://api.weatherapi.com/v1/current.json"

kota = input("Masukkan lokasi yang ingin di periksa: ")

params = {
    'key' : api_key,
    'q' : kota
}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()

    kelembapan = data['current']['humidity']
    kondisi = data['current']['condition']['text']
    suhu = data['current']['temp_c']

    if kelembapan > 80 or "rain" in kondisi.lower():
        print("Peringatan: Potensi hujan tinggi, disarankan membawa payung!")
    elif suhu > 32:
        print("Cuaca cukup terik, pastikan minum air yang cukup!")
    else:
        print("Pakai pakaian yang nyaman, cuaca relatif aman.")
else:
    print(f"Gagal mengambil data. Kota {kota} tidak ditemukan")
````

````python
import requests
import json

api_key = '86412f801a0c4053893194448260208'
url = "http://api.weatherapi.com/v1/current.json"
kota_list = ['Makassar','Jakarta','Bandung']
cuaca = []

for kota in kota_list:
    params = {
        'key': api_key,
        'q': kota
    }

    response = requests.get(url, params=params)

    if response.status_code == 200:
        data = response.json()

        suhu = data['current']['temp_c']
        lokasi = data['location']['name']
        detail = {
            'lokasi' : lokasi,
            'suhu': suhu
        }
        cuaca.append(detail)
    else:
        print('Data tidak ditemukan')
        
print(json.dumps(cuaca, indent=2))

if cuaca:
    terpanas = max(cuaca, key=lambda x: x['suhu'])
    terdingin = min(cuaca, key=lambda x: x['suhu'])

    print("\n=== HASIL KOMPARASI ===")
    print(f" Kota Terpanas : {terpanas['lokasi']} ({terpanas['suhu']}°C)")
    print(f" Kota Terdingin: {terdingin['lokasi']} ({terdingin['suhu']}°C)")
````

````python
import requests
import csv
import json

api_key = '86412f801a0c4053893194448260208'
url = 'http://api.weatherapi.com/v1/current.json'

kota_list = ['Makassar','Surabaya','Manado','Jakarta','Denpasar']
list_data_cuaca = []

for kota in kota_list:
    params = {
        'key': api_key,
        'q': kota
    }

    response = requests.get(url, params=params)
    if response.status_code == 200:
        data = response.json()

        kota = data['location']['name']
        negara = data['location']['country']
        suhu_c = data['current']['temp_c']
        kondisi = data['current']['condition']['text']
        kecepatan_angin = data['current']['wind_kph']
        waktu_lokal = data['location']['localtime']

        dict_data = {
            'Kota': kota,
            'Negara': negara,
            'Suhu Celcius': suhu_c,
            'Kondisi': kondisi,
            'Kecepatan Angin Kwh': kecepatan_angin,
            'Waktu Lokal': waktu_lokal
        }
        list_data_cuaca.append(dict_data)
    else:
        print("Data tidak ditemukan")

with open('laporan_cuaca_harian.csv', 'w', encoding='utf-8') as file:
    header = ['Kota','Negara','Suhu Celcius','Kondisi','Kecepatan Angin Kwh','Waktu Lokal']
    writer = csv.DictWriter(file, fieldnames=header)
    writer.writeheader()
    writer.writerows(list_data_cuaca)
````
